```python
import os
import numpy as np

base = r'C:\Users\neo62\sperm-ai\data\raw\mhsma-dataset-master\mhsma-dataset-master'
mhsma_dir = os.path.join(base, 'mhsma')

print("=== mhsma 폴더 구조 ===")
for root, dirs, files in os.walk(mhsma_dir):
    level = root.replace(mhsma_dir, '').count(os.sep)
    indent = ' ' * 2 * level
    print(f"{indent}{os.path.basename(root)}/")
    subindent = ' ' * 2 * (level + 1)
    for f in files:
        path = os.path.join(root, f)
        size = os.path.getsize(path)
        print(f"{subindent}{f}  ({size/1024:.1f} KB)")

print("\n=== .npy 파일 로드 테스트 ===")
for root, dirs, files in os.walk(mhsma_dir):
    for f in files:
        if f.endswith('.npy'):
            path = os.path.join(root, f)
            data = np.load(path, allow_pickle=True)
            print(f"{f}: shape={data.shape}, dtype={data.dtype}")
```
