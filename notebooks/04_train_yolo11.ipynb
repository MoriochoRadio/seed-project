```python
import os
from collections import Counter

label_dir = r'C:\Users\neo62\sperm-ai\data\processed\yolo_dataset\labels\train'

class_counts = Counter()
for fname in os.listdir(label_dir):
    with open(os.path.join(label_dir, fname), 'r') as f:
        for line in f:
            cls = int(line.strip().split()[0])
            class_counts[cls] += 1

total = sum(class_counts.values())
print("클래스별 분포:")
print(f"  sperm   (0): {class_counts[0]:>7,}개 ({class_counts[0]/total*100:.1f}%)")
print(f"  cluster (1): {class_counts[1]:>7,}개 ({class_counts[1]/total*100:.1f}%)")
print(f"  small   (2): {class_counts[2]:>7,}개 ({class_counts[2]/total*100:.1f}%)")
print(f"  전체        : {total:>7,}개")
```
